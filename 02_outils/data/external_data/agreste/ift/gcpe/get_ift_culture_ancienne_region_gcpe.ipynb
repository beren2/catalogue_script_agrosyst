#### Import des librairies

```python
import pandas as pd
from tqdm import tqdm
import numpy as np
```

```python
path_file = './'

ENTREPOT_PATH = '~/Bureau/utils/data/'
df = {}
```

```python
# campagne : campagne de l'enquête PK
# starting_line : ligne de départ pour la lecture du fichier excel
# values_out_regions : lignes à exclure de la lecture du fichier excel (pour la colonne Région)
# region_column : nom de la colonne où est stockée la région


STUDIES = [
    {
        'campagne' : 2017,
        'starting_line' : 0,
        'values_out_regions' : ['Demi-intervalle de confiance'], 
        'region_column' : 'Région'
    },
    {
        'campagne' : 2021,
        'starting_line' : 1,
        'values_out_regions' : [], 
        'region_column' : ' '
    }
]
```

```python
def load_ift_culture_ancienne_region_gcpe(campagne, start_line=1, values_out_regions=[], region_column='Région'):
    """
        Ouvre le fichier brut des informations Agreste
        et le transforme en un DataFrame pandas avec les colonnes attendues pour la suite :
        - 
        
    """

    file_name = 'ift_culture_ancienne_region_gcpe_brut_' + str(campagne)
    df['ift_culture_ancienne_region_gcpe'] = pd.read_csv(path_file+file_name+'.csv')


    # quelques retructurations sont nécessaires pour faciliter le traitement du fichier
    df['ift_culture_ancienne_region_gcpe'] = df['ift_culture_ancienne_region_gcpe'].iloc[start_line:]
    df['ift_culture_ancienne_region_gcpe'].columns = df['ift_culture_ancienne_region_gcpe'].iloc[0].values
    df['ift_culture_ancienne_region_gcpe'] = df['ift_culture_ancienne_region_gcpe'].iloc[1:-3]
    df['ift_culture_ancienne_region_gcpe'] = df['ift_culture_ancienne_region_gcpe'].rename(columns={
        region_column : 'nom_ancienne_region'
    }).set_index('nom_ancienne_region')

    # on exclue les lignes dont la région vaut les valeurs qu'on veux exclure
    for value in values_out_regions:
        df['ift_culture_ancienne_region_gcpe'] = df['ift_culture_ancienne_region_gcpe'][
            df['ift_culture_ancienne_region_gcpe'].index != value
        ]

    # Convertir les colonnes en float
    df['ift_culture_ancienne_region_gcpe'] = df['ift_culture_ancienne_region_gcpe'].replace(
        to_replace='ns', value=np.nan
    )

    df['ift_culture_ancienne_region_gcpe']  = df['ift_culture_ancienne_region_gcpe'].applymap(
        lambda x: str(x).replace(",", ".") if pd.notna(x) else x
    )

    for col in df['ift_culture_ancienne_region_gcpe'].columns:
        df['ift_culture_ancienne_region_gcpe'][col] = pd.to_numeric(
            df['ift_culture_ancienne_region_gcpe'][col], errors='coerce'
        )
    df['ift_culture_ancienne_region_gcpe'].to_csv('ift_culture_ancienne_region_gcpe_'+str(campagne)+'.csv', index=True)
```

```python
# on applique la fonction pour obtenir le fichier final sur toutes les campagnes étudiées
for study in STUDIES :
    load_ift_culture_ancienne_region_gcpe(
        study['campagne'],
        start_line=study['starting_line'],
        values_out_regions=study['values_out_regions'], 
        region_column=study['region_column']
    )

```
